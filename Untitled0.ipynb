```python
import numpy as np
import matplotlib.pyplot as plt

# Parámetros de tiempo y muestreo
Fs = 1000  # Frecuencia de muestreo (Hz)
T = 1.0    # Duración total (s)
t = np.linspace(0, T, int(Fs * T), endpoint=False)

# 1. Creación de señales elementales
senal_seno = np.sin(2 * np.pi * 50 * t) # Senoidal de 50 Hz

senal_rect = np.zeros_like(t)
senal_rect[(t >= 0.3) & (t <= 0.7)] = 1.0 # Pulso rectangular

senal_escalon = np.ones_like(t)
senal_escalon[t < 0.2] = 0.0 # Función escalón (activada en 0.2s)

# Función para calcular FFT y graficar
def analizar_fft(senal, titulo):
    N = len(senal)
    fft_vals = np.fft.fft(senal)
    fft_freqs = np.fft.fftfreq(N, 1/Fs)

    # Filtrar solo la mitad positiva del espectro
    n_pos = N // 2
    freqs_pos = fft_freqs[:n_pos]
    magnitud = np.abs(fft_vals[:n_pos]) / N
    fase = np.angle(fft_vals[:n_pos])

    plt.figure(figsize=(14, 4))

    # Gráfica en el dominio del tiempo
    plt.subplot(1, 3, 1)
    plt.plot(t, senal, color='blue')
    plt.title(f'Tiempo: {titulo}')
    plt.xlabel('Tiempo (s)')
    plt.ylabel('Amplitud')
    plt.grid(True)

    # Gráfica de Magnitud
    plt.subplot(1, 3, 2)
    plt.plot(freqs_pos, magnitud, color='red')
    plt.title('Espectro de Magnitud')
    plt.xlabel('Frecuencia (Hz)')
    plt.grid(True)

    # Gráfica de Fase
    plt.subplot(1, 3, 3)
    plt.plot(freqs_pos, fase, color='green')
    plt.title('Espectro de Fase')
    plt.xlabel('Frecuencia (Hz)')
    plt.grid(True)

    plt.tight_layout()
    plt.show()

# Ejecutar el análisis para cada señal
analizar_fft(senal_seno, "Señal Senoidal (50 Hz)")
analizar_fft(senal_rect, "Pulso Rectangular")
analizar_fft(senal_escalon, "Función Escalón")
```
